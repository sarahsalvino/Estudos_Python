```python
class triangulo:
    def __init__(self, lado1, lado2, lado3, base, altura):
        self.lado1 = lado1
        self.lado2 = lado2
        self.lado3 = lado3
        self.base = base
        self.altura = altura

    def area(self): 
        return (self.base * self.altura) / 2
    
    def tipo(self):
        if self.lado1 > self.lado2 + self.lado3:
            return 'não é um triângulo'
        elif self.lado1 == self.lado2 or self.lado1 == self.lado3 or self.lado1 == self.lado2:
            return 'triângulo isósceles'
        else:
            return 'outro tipo de triângulo'

t1 = triangulo(2, 1, 3, 4, 3)
print(t1.lado1, t1.lado2, t1.lado3, t1.base, t1.altura)

print(t1.area())  

print(t1.tipo())
```

```python
class Aluno:
    def __init__(self, nome, nota1, nota2):
        self.nome = nome
        self.nota1 = nota1
        self.nota2 = nota2

    def media(self):
        return (self.nota1 + self.nota2) / 2

    def situacao(self):
        if self.media() >= 6:
            return 'aprovado'
        else:
            return 'recuperação'

aluno1 = Aluno('Sarah', 8, 9)
print(aluno1.nome, aluno1.nota1, aluno1.nota2)
print(aluno1.media())
print(aluno1.situacao())

aluno2 = Aluno('Lucas', 4, 5)
print(aluno2.nome, aluno2.nota1, aluno2.nota2)
print(aluno2.media())
print(aluno2.situacao())
```

```python
#Leia duas matrizes 3x3 do tipo inteiro, crie uma terceira matriz que contenha a subtração de A-B, no final, imprima a matriz resultante.

import numpy as np

TAMANHO = 3

matrizP = np.empty([TAMANHO, TAMANHO], dtype=np.int32)
matrizS = np.empty([TAMANHO, TAMANHO], dtype=np.int32)
matrizT = np.empty([TAMANHO, TAMANHO], dtype=np.int32)

print("Leitura da matriz P")
for linha in range(0, TAMANHO):
    for coluna in range(0, TAMANHO):
        print("Digite o valor da linha", linha, "e coluna", coluna, ": ")
        matrizP[linha][coluna] = int(input())


print("Leitura da matriz S")
for linha in range(0, TAMANHO):
    for coluna in range(0, TAMANHO):
        print("Digite o valor da linha", linha, "e coluna", coluna, ": ")
        matrizS[linha] [coluna] = int(input())

print("Matriz resultante")
for linha in range(0, TAMANHO):
    for coluna in range(0, TAMANHO):
        matrizT[linha][coluna] = matrizP[linha][coluna] - matrizS[linha][coluna]
        print(matrizT[linha][coluna]) 
```

```python
#forma corrigida

import numpy as np

TAMANHO = 3
quantidade = 0  # Inicializa a variável
maiores_que_10 = []  # Lista para guardar os números maiores que 10

matriz = np.empty([TAMANHO, TAMANHO], dtype=np.int32)

# Preenchendo a matriz com os valores digitados
for linha in range(0, TAMANHO):
    for coluna in range(0, TAMANHO):
        print("Digite o valor da linha", linha, "e coluna", coluna, ": ")
        matriz[linha][coluna] = int(input())

# Verificando os números maiores que 10
for linha in range(0, TAMANHO):
    for coluna in range(0, TAMANHO):
        if matriz[linha][coluna] > 10:
            quantidade += 1
            maiores_que_10.append(matriz[linha][coluna])

print("Quantidade de números maiores que 10:", quantidade)

if quantidade > 0:
    print("Números maiores que 10:", maiores_que_10)
else:
    print("Não há números maiores que 10.")
```

```python

#Leia uma matriz 2x3 de números inteiros e em seguida:
#Calcule e mostre a média de todos os valores da matriz

import numpy as np

TAMANHO_LINHA = 2
TAMANHO_COLUNA = 3

matriz = np.empty([TAMANHO_LINHA, TAMANHO_COLUNA], dtype=np.int32)

for linha in range(0, TAMANHO_LINHA):
    for coluna in range(0, TAMANHO_COLUNA):
        print("Digite o valor da linha", linha, "e coluna", coluna, ": ")
        matriz[linha][coluna] = int(input())
soma = 0
for linha in range(0, TAMANHO_LINHA):
    for coluna in range(0, TAMANHO_COLUNA):
        soma += matriz[linha][coluna]
media = soma / (TAMANHO_LINHA * TAMANHO_COLUNA)
print("média dos valores é: ", media)       
```

```python
#Crie um vetor de 5 posições, peça ao usuário para preenchê-lo e depois exiba a soma de todos os elementos.

import numpy as np

vetor = np.empty(5, dtype=np.float32)

for posicao in range(0,5):
    print("digite o valor ", posicao + 1, ": ")
    vetor[posicao] = float(input())

    soma = 0 
    for posicao in range(0,5):
        soma = soma + vetor[posicao]
print("A soma dos valores é: ", soma)
```

```python
#Crie um vetor de 8 posições, peça ao usuário para preenchê-lo e depois exiba o maior e o menor valor digitado.

import numpy as np

vetor = np.empty(8, dtype=np.float32)

for posicao in range(0,8):
    print("digite o valor ", posicao + 1, ": ")
    vetor[posicao] = float(input())

for posicao in range(0,8):
    if posicao == 0:
        maior = vetor[posicao]
        menor = vetor[posicao]
    else:
        if vetor[posicao] > maior:
            maior = vetor[posicao]
            print("Esse é o maior: ", maior)
            if vetor[posicao] < menor:
                menor = vetor[posicao]
                print("Esse é o menor: ", menor)
```

```python
#Crie um vetor de 6 posições, peça ao usuário para preenchê-lo e depois exiba a média dos valores digitados.


import numpy as np 
vetor = np.empty(6, dtype=np.float32)

for posicao in range(0,6):
    print("digite o valor", posicao + 1, ":")
    vetor[posicao] = float(input())

    soma = 0
    for posicao in range(0,6):
        soma = soma + vetor[posicao]
        media = soma/6
print("A soma dos valores é: ", soma)        
print("A média dos valores é:", media)     
```

```python
#Crie um vetor de 7 posições, peça ao usuário para preenchê-lo e depois exiba apenas os valores pares.

import numpy as np

vetor = np.empty(7, dtype=np.int32)

for posicao in range(0,7):
    print("digite o valor", posicao + 1, ": ")
    vetor[posicao] = int(input())

    if vetor[posicao] % 2 == 0:
        print("Esse é um valor par: ", vetor[posicao])
    else:
        print("Esse valor é ímpar: ", vetor[posicao])
```
