Exercicio caixa eletronico

```python
saldo = 0

while True:
    print("Saldo atual: R$", saldo)
    opcao = input("Escolha uma opção (depositar, sacar, sair): ").lower()

    if opcao == "depositar":
        valor = float(input("Digite o valor para depositar: "))
        saldo += valor

    elif opcao == "sacar":
        valor = float(input("Digite o valor para sacar: "))
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente!")

    elif opcao == "sair":
        break
    else:
        print("Opção inválida!")

print("Programa encerrado. Seu saldo final é:", saldo)
```

```python
#Leia uma matriz de inteiros 3x3. Após, leia um valor individual e mostre quantas vezes o número digitado aparece na matriz

import numpy as np #importando biblioteca numpy
TAMANHO = 3 #criando uma variável para definir o tamanho da matriz, nesse caso, 3x3

matriz = np.empty([TAMANHO, TAMANHO], dtype=np.int32) #criando uma matriz vazia com o tamanho definido pela variável "TAMANHO" e definindo o tipo de dado como inteiro utilizando a função "empty" da biblioteca numpy


#criação da condição para o usuário inserir os valores da matriz
for linha in range(0,TAMANHO):
    for coluna in range(0,TAMANHO):
        print("digite o valor da linha", linha, "e coluna", coluna, ": ")
        matriz[linha][coluna] = int(input())


print("Digite o valor a ser pesquisado: ")
pesquisa = int(input())

quantidade = 0
for linha in range(0, TAMANHO):
    for coluna in range(0, TAMANHO):
        if matriz[linha][coluna] == pesquisa:
            quantidade += 1
print("quantidade de números encontrados: ", quantidade)

#for linha in range(0,TAMANHO):
    #for coluna in range(0,TAMANHO):
        #print(matriz[linha][coluna])  
```

```python
#Exercicio 4
import numpy as np
vetor = np.empty(10)

for posicao in range(0, 10):
    print("digite o valor", posicao + 1, ": ")
    vetor[posicao] = float(input())

soma = 0.0
negativos = 0

for posicao in range(0, 10):
    soma = soma + vetor[posicao]
    if vetor[posicao] < 0:
        negativos = negativos + 1


media = soma/10
print("soma: ", soma)    
print("média: ", media)
print("negativos: ", negativos)
```
