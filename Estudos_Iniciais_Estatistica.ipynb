```python
#função para gerar os valores da amostragem sistematica

def amostragem_sistematica(dataset, amostras):
    intervalo = int(len(dataset) / amostras) # Passo 1: Calcula de quantas em quantas linhas vamos pegar
    random.seed(1)  # para dar o mesmo resultado sempre que o código rodar
    inicio = random.randint(0, intervalo)  # Passo 2: Sorteia um número pra começar
    indices = np.arange(inicio, len(dataset), step = intervalo)# Passo 3: Cria os índices que vamos usar
    amostra_sistematica = dataset.iloc[indices] # Passo 4: Pega as linhas do dataset com base nesses índices
    return amostra_sistematica 
```

```python
# Função para fazer amostragem por agrupamento

def amostragem_agrupamento(dataset, numero_grupos):
    intervalo = int(len(dataset) / numero_grupos)  #Calcula quantas linhas vão existir em cada grupo
    
    grupos = []  # Lista para armazenar o número do grupo de cada linha
    id_grupo = 0  # ID do grupo atual
    contagem = 0   # Contador de quantas linhas já foram colocadas no grupo atual

# Percorre cada linha do dataset (apenas para contar e agrupar)
    for _ in dataset.iterrows():  
        grupos.append(id_grupo) 
        contagem += 1 # Adiciona o ID do grupo atual na lista
        if contagem >= intervalo: 
            contagem = 0  # Zera o contador
            id_grupo += 1 # Passa para o próximo grupo

    dataset['grupo'] = grupos  # Cria uma nova coluna no dataset com os grupos
    random.seed(1) #para garantir que seja sempre o mesmo
    grupo_selecionado = random.randint(0, numero_grupos - 1) # Sorteia um grupo aleatório entre 0 e o número total de grupos - 1
    return dataset[dataset['grupo'] == grupo_selecionado]   # Retorna apenas as linhas do grupo sorteado
```

```python
def amostragem_reservatorio(dataset, amostras): #criação de função
    tamanho = len(dataset) #referindo o tamanho do dataset para a variável tamanho
    reservatorio = [] #criando lista vazia

    # Passo 1: Preencher o reservatório com os primeiros 'amostras' elementos
    for i in range(amostras): #estrutura de repetição para preencher o reservatorio
        reservatorio.append(i)

    # Passo 2: Para cada elemento depois dos primeiros 'amostras'
    for i in range(amostras, tamanho): #estrutura de repetição para amostraws, e tamanho
        j = random.randint(0, i) #3
        if j < amostras: #se o valor de j for menor que amostras
            reservatorio[j] = i # Substitui um elemento do reservatório por um novo elemento

    # Retornar os elementos sorteados do dataset
    return dataset.iloc[reservatorio]
```

```python

import pandas as pd
import random
import numpy as np


dataset = pd.read_csv(r"C:\Users\SarahCavalcanteSalvi\Desktop\PROJETOS\credit_data.csv")
```

```python
#amostra aleatoria simples

def amostragem_aleatoria_simples(dataset, amostras):
    return dataset.sample(n = amostras)
```

```python
#amostragem sistemática

def amostragem_sistematica(dataset, amostras):
    intervalo = len(dataset) // amostras
    random.seed(1)
    inicio = random.randint(0, intervalo)
    indices = np.arange(inicio, len(dataset), step = intervalo)
    amostra_sistematica = dataset.iloc[indices]
    return amostra_sistematica
```

```python
df_amostra_sistematica = amostragem_sistematica(dataset, 1000)
df_amostra_sistematica.shape
```

```python
#amostragem por grupo

import random

def amostragem_agrupamento(dataset, numero_grupos):
    intervalo = len(dataset) / numero_grupos

    grupos = []
    id_grupo = 0
    contagem = 0

    for _ in dataset.iterrows():
        grupos.append(id_grupo)
        contagem += 1
        if contagem >= intervalo:
            contagem = 0
            id_grupo += 1
            if id_grupo >= numero_grupos:
                id_grupo = numero_grupos - 1  # Garante que não ultrapasse

    dataset['grupo'] = grupos
    grupo_selecionado = random.randint(0, numero_grupos - 1)
    return dataset[dataset['grupo'] == grupo_selecionado]
          
```

```python
df_amostra_agrupamento = amostragem_agrupamento(dataset, 2)
df_amostra_agrupamento.shape, df_amostra_agrupamento['grupo'].value_counts()
```

```python
#amostra estratificada 


from sklearn.model_selection import StratifiedShuffleSplit
```

```python
def amostragem_estratificada(dataset, percentual, campo):
    split = StratifiedShuffleSplit(test_size = percentual, random_state=1)
    for _, y in split.split(dataset, dataset[campo]):
        df_y = dataset.iloc[y]
    return df_y    
```

```python
df_amostra_estratificada = amostragem_estratificada(dataset, 0.5, 'c#default')
df_amostra_estratificada.shape
```

```python
#amostra de reservatorio


def amostragem_reservatorio(dataset, amostras):
    stream = list(range(len(dataset)))
    tamanho = len(dataset)

    reservatorio = [0] * amostras 
    for i in range(amostras):
        reservatorio[i] = stream[i]

    i = amostras
    while i < tamanho:
        j = random.randrange(i + 1)
        if j < amostras:
            reservatorio[j] = stream[i]
        i += 1

    return dataset.iloc[reservatorio]
```

```python
df_amostragem_reservatorio = amostragem_reservatorio(dataset, 1000)
df_amostragem_reservatorio.shape
```
