```python
import numpy as np #Importação da biblioteca numpy

# Criando um DataFrame com 10 linhas e 3 colunas: 'idade', 'peso' e 'nota'.
df = pd.DataFrame({
    'idade': np.random.randint(18, 60, size=10),          #Idades aleatórias entre 18 e 60 anos
    'peso': np.random.uniform(50, 100, size=10).round(1), #Pesos aleatórios entre 50 e 100 kg, arredondados para 1 casa decimal
    'nota': np.random.uniform(0, 10, size=10).round(2)    #Notas aleatórias entre 0 e 10, arredondadas para 2 casas decimais
})

df = df.sort_values(by='idade') #Ordenando o DataFrame pela coluna 'idade' para facilitar a visualização
df.plot(x='idade', y='peso', title='Idade vs Peso')
```

Valores Faltantes

```python
#Valores Faltantes

serie_faltante = pd.Series([1,2,3, np.nan, 5, np.nan]) #Criando uma nova serie contendo valores faltantes
serie_faltante
```

```python
#Verificando se tem valores faltantes .

serie_faltante.isna()
```

```python
#Mostrando a quantidade de valores faltantes. 

serie_faltante.isna().sum()
```

```python
#Mostrando a contagem de valores.

serie_faltante.value_counts(dropna = False)
```

```python
#Significa que os valores faltantes serão iguais a 0.

serie_faltante.fillna(0)
```

```python
#Apaga os valores faltantes.

serie_faltante.dropna()
```

```python
#Extraindo media das series.

serie_faltante.fillna(serie_faltante.mean())
```

```python
#Criando uma nova série com valores faltantes e string.

serie_faltante = pd.Series(["Maça", "Banana", "Arroz", "Arroz", np.nan, "Batata"])
serie_faltante
```

```python
#Mostrando a quantidade de valores faltantes.

serie_faltante.isna().sum()
```

```python
#Substitui os valores faltantes por "Não informado".

serie_faltante.fillna("Não informado")
```

```python
#Mostrando o valor mais frequente da serie (moda).

serie_faltante.mode().iloc[0]
```

```python
#Substitui o valor faltante pelo mais frequente (moda) da serie.

serie_faltante.fillna(serie_faltante.mode().iloc[0])
```

Funções 

```python
#Funções 

serie_idade
```

```python
#Mostrando os valores menores que 18.

serie_idade.loc[serie_idade < 18] 
```

```python
#Criando função onde vai transformar os valores menores que 18 em 18.

def corrige_idade(idade):
    if idade < 18:
        idade = 18
    return idade
```

```python
corrige_idade(15) #Conferindo a função criada.
```

```python
#Aplicando a função na serie_idade.

serie_idade = serie_idade.apply(corrige_idade)
```

```python
#Lambda = forma de criar uma função "curta" nesse caso se x for igual a 18, ele vai ser 17.

serie_idade = serie_idade.apply(lambda x: 17 if x == 18 else x) 
```

```python
#Mostrando os valores menores que 18.

serie_idade.loc[serie_idade < 18]
```
